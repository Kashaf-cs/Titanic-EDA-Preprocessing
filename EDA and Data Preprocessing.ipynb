**Check Missing Values ✅**

```python
df.isnull().sum()
```

**Observation:**

The dataset contains missing values in three columns. The Age column has 177 missing values, Cabin has 687 missing values, and Embarked has 2 missing values. All other columns have no missing values.

**Mean, Median and Mode ✅**

```python
df["Age"].mean()
```

```python
df["Age"].median()
```

```python
df["Age"].mode()
```

```python
sns.histplot(df["Age"], kde=True)

plt.axvline(df["Age"].mean(), color="red", label="Mean")
plt.axvline(df["Age"].median(), color="green", label="Median")
plt.axvline(df["Age"].mode()[0], color="blue", label="Mode")

plt.title("Age Distribution with Mean, Median and Mode")
plt.legend()
plt.show()
```

**Normal and Skewed Distribution ✅**

```python
sns.histplot(df["Age"], kde=True)
plt.show()
```

**Observation:**

Age is approximately normally distributed.

 **Skewed Distribution**

```python
sns.histplot(df['Fare'], kde=True)
plt.show()
```

**Observation:**

The Fare column is positively (right) skewed. Most passengers paid lower fares, while a small number of passengers paid very high fares, creating a long tail on the right side of the distribution.

**Box Plot (Outlier Detection) ✅**

```python
sns.boxplot(x=df['Fare'])
plt.show()
```

**Observation:**

The Fare column contains many outliers. Most passengers paid low fares, while a few passengers paid very high fares. These extreme values appear as points outside the whiskers of the box plot.

**Binning ✅**

```python
bins=[0,18,35,60,100]

labels=["Child","Young Adult","Adult","Senior"]

df["Age_Group"]=pd.cut(df["Age"],bins=bins,labels=labels)

df[["Age","Age_Group"]].head()
```

**Observation:**

The Age column was divided into four groups using binning: Child (0–18), Young Adult (18–35), Adult (35–60), and Senior (60–100). Binning makes continuous numerical data easier to understand and analyze by converting it into meaningful categories.

**Different Data Visualization Charts**

**Histogram ✅**

```python
sns.histplot(df['Age'] ,kde=True)
plt.show()
```

**Observation:**

The histogram shows that most passengers are between 20 and 35 years of age. There are fewer very young and very old passengers. The Age distribution is slightly right-skewed.

**Bar Chart ✅**

```python
sns.countplot(x="Sex",data=df)
plt.show()
```

**Observation:**

The bar chart shows that the number of male passengers is higher than the number of female passengers in the dataset.

**Scatter Plot ✅**

```python
sns.scatterplot(x="Age",y="Fare",data=df)
plt.show()
```

**Observation:**

The scatter plot shows the relationship between Age and Fare. Most passengers paid fares below 50, while a few passengers paid very high fares, indicating outliers. There is no strong relationship between Age and Fare because the points are widely scattered.

**Line Chart ✅**

```python
df.groupby("Pclass")["Fare"].mean().plot(kind="line")
plt.show()
```

**Observation:**

First-class passengers paid the highest average ticket fare, second-class passengers paid a moderate fare, and third-class passengers paid the lowest fare. This shows that higher passenger class is associated with higher ticket prices

**Heatmap ✅**

```python
corr=df.corr(numeric_only=True)

sns.heatmap(corr,annot=True,cmap="coolwarm")
```

**Observation:**

The heatmap displays the correlation coefficients between the numerical features in the dataset. A strong negative correlation is observed between Pclass and Fare (-0.55), indicating that passengers in higher classes (where Pclass = 1) generally paid higher fares. Pclass also has a negative correlation with Survived (-0.34), while Fare has a positive correlation with Survived (0.26), suggesting that passengers who traveled in higher classes and paid higher fares had a greater likelihood of survival. Additionally, SibSp and Parch show a moderate positive correlation (0.41), indicating that passengers traveling with siblings/spouses were also likely to be traveling with parents or children
